Store Passenger Data Using Lists



```python
passengers = ["Rahul Sharma", "Priya Patel", "Amit Verma"]

print("Passenger Reservation List:")
print(passengers)

print("\nFirst Passenger:", passengers[0])

print("Second Passenger:", passengers[1])

print("Third Passenger:", passengers[2])
```

 List Operations



Create a List

```python
passengers = ["Rahul Sharma", "Priya Patel", "Amit Verma"]
```

Display Original List

```python
print(passengers)
```

Add New Passenger Using append()

```python
passengers.append("Sneha Joshi")
print(passengers)
```

Remove Passenger Using remove()

```python
passengers.remove("Amit Verma")
print(passengers)
```

Insert Passenger Using insert()

```python
passengers.insert(1, "Rohan Mehta")
print(passengers)
```

Remove Item Using pop()

```python
removed_passengers = passengers.pop(2)
print(passengers)
```

Display Updated List

```python
print("\nUpdated Reservation List:")
print(passengers)
```
