Lambda Function for Fare Discount

```python
discount_fare = lambda fare: fare - 200
```

```python
print("Discounted Fare:", discount_fare(2500))  # Output: Discounted Fare: 800
```

map() Function for Fare Calculation


```python
fare = [2500, 300, 1560, 4000]
updated_fares = list(map(lambda fare: fare + 100, fare))
```

```python
print(updated_fares)  # Output: [2600, 3100, 1600, 4100]
```

filter() Function for confirmed tickets



```python
reservation_status = [
    "Confirmed",
    "Waiting",
    "Confirmed",
    "RAC"
]
```

```python
confirmed_tickets = list(
    filter(
        lambda status: status == "Confirmed", reservation_status
    )
)
```

```python
print(confirmed_tickets)
```

Combine Passenger and Seat Data Using zip()

```python
passengers = [
    "Rahul Sharma",
    "Priya Singh",
    "Amit Verma"
]

seats = [
    "B1-21",
    "B1-22",
    "B1-23"
]
```

```python
reservation_data = list(
    zip(passengers, seats)
)
```

```python
print(reservation_data)
```

enumerate() Function

```python
passenger_list = [
    "Rahul Sharma",
    "Priya Singh",
    "Amit Verma"
]
```

```python
for index, passenger in enumerate(passenger_list):
    print(index, passenger)
```

Process Data Using map() and filter()

```python
ticket_fares = [2500, 800, 3200, 600]
```

```python
valid_fares = list(
    filter(
        lambda fare: fare > 1000, ticket_fares
    )
)
```

```python
final_fares = list(
    map(
        lambda fare: fare + 100, valid_fares
    )
)
```

```python
print(final_fares)
```

Final Functional Reservation Module



```python
passengers = ["Rahul", "Priya", "Amit"]
seats = ["S1", "S2", "S3"]
fares = [2500, 1800, 3200]
```

```python
reservation_data = list(
    zip(passengers, seats, fares)
)
```

```python
print("=============================================")
print(" Railway Reservation Report ")
print("=============================================")

for index, data in enumerate(reservation_data):

    passenger, seat, fare = data

    final_fare =  (lambda x: x - 100) (fare)

    print("Reservation Id :", index + 1)
    print("Passenger :", passenger)
    print("Seat :", seat)
    print("Final Fare :", final_fare)
    print("---------------------------------------------")
```
