Store Train Data Using Dictionaries


```python
train = {
    "Train_Number": 12951,
    "Train_Name": "Mumbai Rajdhani Express",
    "Source": "Mumbai",
    "Destination": "Delhi",
    "Available_Seats": 120
}
```

```python
print("Train Information:")
print(train)

print("\nTrain Number:", train["Train_Number"])
print("Train Name:", train["Train_Name"])
print("Source Station:", train["Source"])
print("Destination Station:", train["Destination"])
print("Available_Seats:", train["Available_Seats"])
```

Access and Update Values

```python
train = {
    "Train_Number": 12951,
    "Train_Name": "Mumbai Rajdhani Express",
    "Available_Seats": 120
}
```

```python
print(train)
```

Access Dictionary Values

```python
print("Train Name:", train["Train_Name"])
print("Available Seats:", train["Available_Seats"])
```

Update Seat Availability


```python
train["Available_Seats"] = 95
print(train)
```

Add New Value Using update()


```python
train.update({"Platform": 4})
print(train)
```

Remove Value Using pop()


```python
train.pop("Platform")
print(train)
```

Display Updated Dictionary

```python
print("\nUpdated train Record:")
print(train)
```
